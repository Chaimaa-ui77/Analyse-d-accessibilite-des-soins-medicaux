```python
import pandas as pd
import openpyxl
```

```python
path_densite = (
    "https://minio.lab.sspcloud.fr/mangane/Projet_PDS_2026/raw/fichier_diffusion_2025grille%20densit%C3%A9.xlsx"
)

densite_file = pd.ExcelFile(path_densite)

print(densite_file.sheet_names)
```

```python
doc_densite = pd.read_excel(
    path_densite,
    sheet_name="Documentation",
    header=None
)

print(doc_densite.head(30).to_string())
```

```python
df_densite = pd.read_excel(
    path_densite,
    sheet_name="Maille communale",
    header=4,
    dtype={"CODGEO": str}
)
```

```python
df_densite = df_densite[
    [
        "CODGEO",
        "LIBGEO",
        "DENS",
        "LIBDENS",
        "DENS7",
        "LIBDENS7"
    ]
].copy()
```

```python
print("Codes communes manquants :", df_densite["CODGEO"].isna().sum())
print("Doublons CODGEO :", df_densite["CODGEO"].duplicated().sum())

print(df_densite.head())
print(df_densite.shape)
```

```python
path_densite_clean = (
    "s3://mangane/Projet_PDS_2026/processed/densite_communes_2025.csv"
)

df_densite.to_csv(
    path_densite_clean,
    index=False
)
```

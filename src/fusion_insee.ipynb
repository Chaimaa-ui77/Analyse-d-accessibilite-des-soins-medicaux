```python
import pandas as pd

import openpyxl
```

```python
path_population = "https://minio.lab.sspcloud.fr/mangane/Projet_PDS_2026/raw/TD_POP1A_2022.xlsx"
path_filosofi = "https://minio.lab.sspcloud.fr/mangane/Projet_PDS_2026/raw/base-cc-filosofi-2021-geo2024.xlsx"
path_densite = "https://minio.lab.sspcloud.fr/mangane/Projet_PDS_2026/raw/fichier_diffusion_2025grille%20densit%C3%A9.xlsx"

df_population = pd.read_csv(path_population, dtype={"CODGEO": str})
df_filosofi = pd.read_csv(path_filosofi, dtype={"CODGEO": str})
df_densite = pd.read_csv(path_densite, dtype={"CODGEO": str})
```

```python
print(df_population.shape)
print(df_filosofi.shape)
print(df_densite.shape)
```

```python
codes_population = set(df_population["CODGEO"])
codes_filosofi = set(df_filosofi["CODGEO"])
codes_densite = set(df_densite["CODGEO"])
```

```python
print("Dans population mais pas dans Filosofi :", len(codes_population - codes_filosofi))
print("Dans Filosofi mais pas dans population :", len(codes_filosofi - codes_population))

print("Dans population mais pas dans densité :", len(codes_population - codes_densite))
print("Dans densité mais pas dans population :", len(codes_densite - codes_population))
```

```python
diff_pop_filo = df_population[
    ~df_population["CODGEO"].isin(df_filosofi["CODGEO"])
]

print(diff_pop_filo[["CODGEO", "LIBGEO"]].head(20))

```

```python
diff_filo_pop = df_filosofi[
    ~df_filosofi["CODGEO"].isin(df_population["CODGEO"])
]

print(diff_filo_pop[["CODGEO", "LIBGEO"]].head(30))
```

```python
diff_dens_pop = df_densite[
    ~df_densite["CODGEO"].isin(df_population["CODGEO"])
]

print(diff_dens_pop[["CODGEO", "LIBGEO"]])
```

```python
df_insee = df_population.merge(
    df_densite,
    on=["CODGEO", "LIBGEO"],
    how="left"
)

print(df_insee.head())
print(df_insee.shape)
```

## Harmonisation de Filosofi vers la géographie 2025

```python
path_passage = "https://minio.lab.sspcloud.fr/mangane/Projet_PDS_2026/raw/table_passage_annuelle_2025.xlsx"

passage_file = pd.ExcelFile(path_passage)

print(passage_file.sheet_names)
```

```python
variables_passage = pd.read_excel(
    path_passage,
    sheet_name="Variables",
    header=None
)

print(variables_passage.head(30).to_string())
```

```python
df_passage = pd.read_excel(
    path_passage,
    sheet_name="COM",
    header=5,
    dtype={
        "CODGEO_2024": str,
        "CODGEO_2025": str
    }
)

df_passage = df_passage[
    df_passage["NIVGEO"] == "COM"
].copy()
```

```python
df_passage = df_passage[
    [
        "CODGEO_2024",
        "LIBGEO_2024",
        "CODGEO_2025",
        "LIBGEO_2025"
    ]
].copy()
```

```python
changements = df_passage[
    (df_passage["CODGEO_2024"] != df_passage["CODGEO_2025"])
    |
    (df_passage["LIBGEO_2024"] != df_passage["LIBGEO_2025"])
]

print(changements.head(30))
print(changements.shape)
```

```python
changements = changements.drop_duplicates(
    subset=["CODGEO_2024", "CODGEO_2025"]
)

print(changements.shape)
```

```python
df_changements_code = pd.read_excel(
    path_passage,
    sheet_name="Liste des changements de code",
    header=5,
    dtype={
        "COM_INI": str,
        "COM_FIN": str
    }
)
```

```python
df_changements_code_2025 = df_changements_code[
    df_changements_code["ANNEE_MODIF"] == 2025
].copy()
```

```python
print(df_changements_code_2025)
print(df_changements_code_2025.shape)
```

```python
df_filosofi_2025 = df_filosofi.copy()
df_filosofi_2025["CODGEO_2024"] = df_filosofi_2025["CODGEO"]


mapping_codes = {
    "12076": "12218",  # Conques-en-Rouergue
    "14011": "14581",  # Aurseulles
    "49069": "49126",  # Orée d'Anjou
    "69159": "69114"   # Porte des Pierres Dorées
}


df_filosofi_2025["CODGEO"] = (
    df_filosofi_2025["CODGEO"]
    .astype(str)
    .replace(mapping_codes)
)
```

```python
df_fusions = pd.read_excel(
    path_passage,
    sheet_name="Liste des fusions",
    header=5,
    dtype={"COM_INI": str, "COM_FIN": str}
)

df_scissions = pd.read_excel(
    path_passage,
    sheet_name="Liste des scissions",
    header=5,
    dtype={"COM_INI": str, "COM_FIN": str}
)

codes_fusions = set(
    df_fusions.loc[
        df_fusions["ANNEE_MODIF"] == 2025,
        "COM_INI"
    ]
)

codes_scissions = set(
    df_scissions.loc[
        df_scissions["ANNEE_MODIF"] == 2025,
        "COM_INI"
    ]
)

codes_non_comparables = codes_fusions | codes_scissions

masque = df_filosofi_2025["CODGEO_2024"].isin(codes_non_comparables)

df_filosofi_2025.loc[
    masque,
    ["MED21", "TP6021", "RD21"]
] = pd.NA
```

```python
df_filosofi_2025 = df_filosofi.copy()
```

```python
df_final = df_insee.merge(
    df_filosofi_2025[
        ["CODGEO", "MED21", "TP6021", "RD21"]
    ],
    on="CODGEO",
    how="left"
)
```

```python
print(df_final.shape)

print("Doublons :", df_final["CODGEO"].duplicated().sum())

print("\nValeurs manquantes :")
print(df_final[["MED21", "TP6021", "RD21"]].isna().sum())
```

```python
path_final = (
    "s3://mangane/Projet_PDS_2026/processed/insee_communes_final.csv"
)

df_final.to_csv(
    path_final,
    index=False
)
```
